**Renewable Energy Percentage**

```python
total_energy = df['total_renewable_energy'].sum()

wind_percentage = (df['wind_energy'].sum() / total_energy) *100

solar_percentage =(df['solar_energy'].sum() / total_energy) *100

other_percentage = (df['other_renewable_energy'].sum() /total_energy) *100


print("wind contribution:", round(wind_percentage,2), "%" )
print("solar contribution:", round(solar_percentage,2), "%" )
print("other renewable contribution:", round(other_percentage,2), "%" )
```

**Observation**

Solar Energy: Solar energy contributes **53.41%** of the total renewable energy, making it the largest contributor.

Wind Energy: Wind energy contributes ** 39.96%,** making it the second-largest contributor.

Other Renewable Energy: Other renewable energy sources contribute **6.66%,** which is the lowest share.

**Overall Observation:**
Solar and wind energy together account for approximately 93.37% of the total renewable energy.

**Identify the highest and lowest energy-generating state**

**Monthly Generation Pattern**

```python
import pandas as pd
import matplotlib.pyplot as plt

df['date'] = pd.to_datetime(df['date'])

df['Month'] = df['date'].dt.month

monthly = df.groupby('Month')['total_renewable_energy'].sum()

print(monthly)

monthly.plot(kind='bar', figsize=(10,5))
plt.title('Monthly Renewable Energy Generation')
plt.xlabel('Month')
plt.ylabel('Total Renewable Energy')
plt.show()
```

**Observation**

The monthly analysis shows fluctuations in renewable energy generation throughout the year. Generation was highest during July and lowest during October and November. These variations may be associated with seasonal changes and differences in energy-source availability.

```python
df['Year'] = df['date'].dt.year

yearly = df.groupby('Year')['total_renewable_energy'].sum()

print(yearly)

yearly.plot(kind='line', marker='o', figsize=(10,5))
plt.title('Yearly Renewable Energy Generation (2019–2025)')
plt.xlabel('Year')
plt.ylabel('Total Renewable Energy')
plt.grid(True)
plt.show()
```

**Conclusion**

Overall, the analysis indicates substantial growth in renewable energy generation during the period from 2019 to 2024.

**Outliers**

```python
energy_columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

for col in energy_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)
    ]

    print(f"\n{col}")
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outlier Count:", len(outliers))
```

```python
import matplotlib.pyplot as plt

df[energy_columns].plot(
    kind='box',
    subplots=True,
    layout=(2, 2),
    figsize=(12, 8)
)

plt.suptitle('Outlier Analysis of Renewable Energy Sources')
plt.tight_layout()
plt.show()
```

**Interpretation Outlier Analysis**

**Wind Energy:** The box plot shows a large number of outliers above the upper whisker. This indicates that some observations have significantly higher wind energy values than the majority of the data.

**Solar Energy:** The box plot also contains many upper outliers, indicating that solar energy values vary considerably, with some observations recording exceptionally high energy generation.

**Other Renewable Energy:** The box plot shows numerous outliers, including some negative values and a few exceptionally high values above 100. This indicates significant variation in other renewable energy generation data. Negative values should be checked for possible data-quality issues.

**Total Renewable Energy:** The box plot shows a large number of upper outliers, with some values exceeding 1,000. This indicates that total renewable energy generation varies considerably across observations, with certain periods or regions recording much higher generation.

**Overall Conclusion:** All four energy variables — wind, solar, other renewable, and total renewable energy — contain numerous outliers. These may reflect genuine differences in energy generation across states and time periods.

**Kurtosis Analysis**

```python
energy_columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

print(df[energy_columns].kurtosis())
```

```python
import matplotlib.pyplot as plt

energy_columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

kurtosis_values = df[energy_columns].kurtosis()

plt.figure(figsize=(8,5))
plt.bar(kurtosis_values.index, kurtosis_values.values)

plt.title("Kurtosis of Renewable Energy Sources")
plt.xlabel("Energy Sources")
plt.ylabel("Kurtosis")
plt.xticks(rotation=20)
plt.axhline(0, linestyle="--")


for i, value in enumerate(kurtosis_values.values):
    plt.text(i, value, f"{value:.2f}", ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import kurtosis

columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

plt.figure(figsize=(10, 6))

for col in columns:
    sns.kdeplot(
        data=df[col].dropna(),
        label=f"{col} (Kurtosis: {kurtosis(df[col].dropna()):.2f})"
    )

plt.title("Kurtosis Distribution of Renewable Energy Sources")
plt.xlabel("Energy Generation")
plt.ylabel("Density")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretation**

Wind Energy: Wind energy has a kurtosis value of 60.54, indicating a highly peaked distribution with potential extreme values or outliers.

Solar Energy: Solar energy has a kurtosis value of 36.50, suggesting a highly peaked distribution with potential extreme values.

Other Renewable Energy: Other renewable energy has a kurtosis value of 31.93, indicating a highly peaked distribution and possible outliers.

Overall Observation: All displayed energy sources have high positive kurtosis values, suggesting that their distributions may contain heavy tails and extreme observations.

```python
import matplotlib.pyplot as plt

energy_columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

skewness_result = df[energy_columns].skew()

plt.figure(figsize=(10, 6))
plt.bar(skewness_result.index, skewness_result.values)

plt.title('Skewness Analysis of Renewable Energy')
plt.xlabel('Energy Sources')
plt.ylabel('Skewness Value')
plt.axhline(y=0, color='black', linestyle='--')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

columns = [
    'wind_energy',
    'solar_energy',
    'other_renewable_energy',
    'total_renewable_energy'
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, col in zip(axes.flatten(), columns):
    values = df[col].dropna()

    sns.histplot(
        values,
        bins=50,
        stat="density",
        kde=True,
        ax=ax
    )

    skew_value = skew(values)

    ax.set_title(f"{col}\nSkewness = {skew_value:.2f}")
    ax.set_xlabel("Energy Generation")
    ax.set_ylabel("Density")

plt.suptitle("Skewness Distribution of Renewable Energy Sources")
plt.tight_layout()
plt.show()
```

**Interpretation**

Wind Energy: The skewness value is approximately 6.9, indicating a highly positively skewed distribution.

Solar Energy: The skewness value is approximately 5.6, showing a positively skewed distribution.

Other Renewable Energy: The skewness value is approximately 5.1, indicating a positively skewed distribution.

Total Renewable Energy: The skewness value is approximately 5.5, showing a positively skewed distribution.

Overall Observation: All energy variables are positively skewed, suggesting that most observations have relatively lower values while a few observations have very high values.

**Conclusion**

The skewness analysis indicates that renewable energy data is not symmetrically distributed. Wind energy shows the highest positive skewness among the displayed variables, suggesting a stronger right tail. Further analysis using box plots can help identify potential outliers.
